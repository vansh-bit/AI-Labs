This notebook implements and investigates autoregressive language models formulated as Bayesian networks.

## 1. Dataset Preparation
We prepare the 6 training sentences with `<START>` and `<END>` tokens.

```python
from autoregressive_bn import prepare_tokenized_dataset, RAW_TRAINING_CORPUS

dataset = prepare_tokenized_dataset(RAW_TRAINING_CORPUS)
for i, sent in enumerate(dataset, 1):
    print(f'{i}: {" ".join(sent)}')
```

## 2. First-Order Autoregressive Bayesian Network
Estimating $P(X_t \mid X_{t-1})$ from bigram transition counts.

```python
from autoregressive_bn import FirstOrderLanguageModel, verify_probability_invariants, START_TOKEN

lm1 = FirstOrderLanguageModel()
lm1.train(dataset)
verify_probability_invariants(lm1)

target_words = [START_TOKEN, 'the', 'cat', 'dog', 'sat', 'ran', 'on', 'to', 'mat', 'rug', 'park']
for w in target_words:
    d = lm1.get_distribution(w)
    print(f'{w:8s} -> {d}')
```

## 3. Next-Word Prediction (Argmax) & Generation
Comparing deterministic greedy decoding against probabilistic sampling.

```python
print('--- Mode A: Greedy Generation ---')
print(' '.join(lm1.generate_sentence(mode='greedy')))

print('\n--- Mode B: Probabilistic Sampling (5 runs) ---')
for r in range(1, 6):
    print(f'Run {r}: {" ".join(lm1.generate_sentence(mode="sample", seed=r * 42))}')
```

## 4. Second-Order Autoregressive Bayesian Network
Conditioning on two preceding tokens $P(X_t \mid X_{t-2}, X_{t-1})$.

```python
from autoregressive_bn import SecondOrderLanguageModel, verify_second_order_invariants

lm2 = SecondOrderLanguageModel()
lm2.train(dataset)
verify_second_order_invariants(lm2)

print('--- Second-Order Greedy Generation ---')
print(' '.join(lm2.generate_sentence(mode='greedy')))

print('\n--- Second-Order Sampling (5 runs) ---')
for r in range(1, 6):
    print(f'Run {r}: {" ".join(lm2.generate_sentence(mode="sample", seed=r * 42))}')
```
