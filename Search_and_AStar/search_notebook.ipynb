**Course:** Artificial Intelligence (CS F407)
**Topic:** Warehouse Robot Navigation using A* and BFS

This notebook implements problem formulation, A* with multiple heuristics, BFS comparison, and edge-case testing.

```python
import math, heapq
from collections import deque
from search_agent import SearchProblem, a_star_search, breadth_first_search, heuristic_manhattan, heuristic_euclidean, heuristic_zero, heuristic_overestimated, ORIGINAL_WAREHOUSE_MAP

problem = SearchProblem(ORIGINAL_WAREHOUSE_MAP)
print(f"Loaded warehouse: {problem.rows}x{problem.cols}, Start: {problem.start}, Goal: {problem.goal}")
```

```python
# Run A* and BFS on original warehouse
res_astar = a_star_search(problem, heuristic_manhattan)
res_bfs = breadth_first_search(problem)

print("--- Benchmark on Original Warehouse ---")
print(f"A*  Path Length: {res_astar.length}, States Expanded: {res_astar.states_expanded}")
print(f"BFS Path Length: {res_bfs.length}, States Expanded: {res_bfs.states_expanded}")
```

```python
# Heuristic Investigation
heuristics = [
    ("h(n) = 0 (Uniform Cost)", heuristic_zero),
    ("h(n) = Euclidean", heuristic_euclidean),
    ("h(n) = Manhattan", heuristic_manhattan),
    ("h(n) = 2 * Manhattan", heuristic_overestimated)
]
for name, fn in heuristics:
    res = a_star_search(problem, fn)
    print(f"{name:25s} -> Found: {res.found}, Length: {res.length}, Expanded: {res.states_expanded}")
```
